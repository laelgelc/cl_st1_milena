## Load AO3 works metadata with stable ID types

```python
from pathlib import Path

import pandas as pd

ao3_works_metadata_path = Path("corpus/00_sources/ao3_works_metadata/ao3_works_metadata.ndjson")

ao3_works_metadata_df = pd.read_json(
    ao3_works_metadata_path,
    lines=True,
    dtype={
        "work_id": "string",
    },
)

ao3_works_metadata_df
```

## Save AO3 work collection plan

```python
from pathlib import Path

work_collection_plan_dir = Path("corpus/00_sources/ao3_work_collection_plan")
work_collection_plan_dir.mkdir(parents=True, exist_ok=True)

work_collection_plan_base = work_collection_plan_dir / "ao3_work_collection_plan"

ao3_works_metadata_df.to_json(
    work_collection_plan_base.with_suffix(".ndjson"),
    orient="records",
    lines=True,
    force_ascii=False,
)

ao3_works_metadata_df.to_csv(
    work_collection_plan_base.with_suffix(".tsv"),
    sep="\t",
    index=False,
)

ao3_works_metadata_df.to_excel(
    work_collection_plan_base.with_suffix(".xlsx"),
    index=False,
)

print(f"Saved {len(ao3_works_metadata_df):,} records to {work_collection_plan_dir}")
```
